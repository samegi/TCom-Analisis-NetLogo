# Análisis de Resultados Finales — Modelo Virus (NetLogo)

Este notebook analiza los resultados del experimento **BehaviorSpace** del modelo *Virus* de NetLogo. Se explora cómo varía el número de tortugas enfermas al final de la simulación según el nivel de infectividad del virus.

## 1. Cargar tidyverse

Cargamos la librería `tidyverse`, que incluye `ggplot2` para visualización y `dplyr` para manipulación de datos.

```r
library(tidyverse)
```

## 2. Leer el CSV final

Cargamos el archivo CSV exportado desde BehaviorSpace. Los primeros 6 renglones contienen metadatos del experimento, por eso usamos `skip = 6`.

```r
url <- "https://raw.githubusercontent.com/samegi/TCom-Analisis-NetLogo/main/virus_infectiousness_final.csv"
data <- read.csv(url, encoding = "UTF-8", skip = 6)
```

## 3. Inspeccionar columnas

Revisamos la estructura del dataset: nombres de columnas, tipos de datos y las primeras filas para entender qué registró el experimento.

```r
head(data)
names(data)
glimpse(data)
```

## 4. Renombrar las columnas

Los nombres originales exportados por NetLogo incluyen caracteres especiales. Los reemplazamos por nombres simples en español para facilitar el análisis:

| Columna original | Nuevo nombre | Descripción |
|---|---|---|
| X.run.number. | corrida | Número de repetición del experimento |
| infectiousness | infectividad | Valor del parámetro manipulado |
| X.step. | paso | Tick final de la simulación |
| count.turtles.with..sick.. | enfermos | Métrica final: tortugas enfermas |

```r
colnames(data) <- c("corrida", "infectividad", "paso", "enfermos")
```

## 5. Seleccionar el parámetro y la métrica final

Filtramos solo las dos columnas relevantes para el análisis: el parámetro manipulado (`infectividad`) y la métrica de resultado (`enfermos`).

```r
datas <- data %>%
  select(infectividad, enfermos)
```

## 6. Gráfica de dispersión

Una primera visualización básica para ver la relación entre infectividad y número de enfermos al final de cada corrida. Cada punto representa una simulación.

```r
datas %>%
  ggplot(aes(infectividad, enfermos)) +
  geom_point()
```

## 7-8. Dispersión con jitter y color por parámetro

Cuando varias corridas comparten el mismo valor de infectividad, los puntos se superponen. El `geom_jitter` los desplaza ligeramente para que sean visibles. El color diferencia visualmente cada nivel del parámetro.

```r
datas %>%
  ggplot(aes(infectividad, enfermos)) +
  geom_point(aes(color = infectividad)) +
  geom_jitter(aes(color = infectividad))
```

## 9. Boxplot agrupado por parámetro

El boxplot resume la distribución de resultados para cada valor de infectividad. Permite comparar mediana, variabilidad y valores atípicos entre los diferentes niveles del parámetro.

```r
datas %>%
  ggplot(aes(infectividad, enfermos)) +
  geom_jitter(aes(color = infectividad)) +
  geom_boxplot(aes(group = infectividad, color = infectividad), alpha = 0.3)
```

## 10. Interpretación de los resultados

### ¿Qué muestra el boxplot?

El experimento corrió múltiples simulaciones con distintos valores del parámetro `infectividad`. El boxplot permite observar:

- **Mediana**: el valor típico de tortugas enfermas al final de la simulación para cada nivel de infectividad.
- **Caja (IQR)**: el rango donde se concentra el 50% de las corridas, indicando qué tan consistente es el modelo.
- **Bigotes**: la variabilidad total de los resultados.
- **Puntos atípicos**: corridas donde el resultado fue inusualmente alto o bajo.

### Relación infectividad → enfermos

En general, a mayor infectividad, se espera un mayor número de tortugas enfermas al final. El boxplot muestra en su mayoria medianas crecientes con el parámetro, por ende el modelo captura correctamente este comportamiento.

### Variabilidad entre corridas

La grafica muestre que las cajas del boxplot son amplias, esto indica que el resultado final depende fuertemente de la estocasticidad del modelo (azar en los movimientos y contactos).
